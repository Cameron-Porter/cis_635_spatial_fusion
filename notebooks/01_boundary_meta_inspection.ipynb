# 1. Inspect geoBoundaries ADM2 metadata

This notebook inspects the source metadata before downloading boundary data. The goal is to document provenance: who published the boundary file, what year it represents, how many ADM2 units it contains, and where the GeoJSON can be retrieved.


## Load project paths and request metadata

The metadata endpoint is kept as a reusable constant in `src/childreach/paths.py`, while the notebook remains independently runnable.


```python
from pathlib import Path
import sys

NOTEBOOK_DIR = Path.cwd()
PROJECT_ROOT = NOTEBOOK_DIR.parent if NOTEBOOK_DIR.name == "notebooks" else NOTEBOOK_DIR
SRC_DIR = PROJECT_ROOT / "src"
if str(SRC_DIR) not in sys.path:
    sys.path.append(str(SRC_DIR))

import requests

from childreach.paths import GEBOUNDARIES_ADM2_METADATA_URL

response = requests.get(GEBOUNDARIES_ADM2_METADATA_URL, timeout=30)
response.raise_for_status()

metadata = response.json()

print("Available metadata fields")
for key in metadata.keys():
    print("-", key)
```

## Print the fields needed for provenance

These are the fields we care about for the README, Obsidian notes, and final methods section.


```python
fields = [
    "admUnitCount",
    "boundaryID",
    "boundaryName",
    "boundaryISO",
    "boundaryType",
    "boundaryYearRepresented",
    "boundarySource",
    "boundarySourceURL",
    "boundaryLicense",
    "gjDownloadURL",
    "simplifiedGeometryGeoJSON",
]

for field in fields:
    print(f"{field}: {metadata.get(field)}")
```

## Interpretation

For Milestone 1, this metadata supports using Mozambique ADM2 boundaries as the district/district-like analysis unit. Boundary choice affects every downstream statistic, so the source, year, license, and feature count should remain documented.
