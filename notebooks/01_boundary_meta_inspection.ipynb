```python
import requests

url = "https://www.geoboundaries.org/api/current/gbOpen/MOZ/ADM2/"
response = requests.get(url)
response.raise_for_status()

metadata = response.json()

for key in metadata.keys():
    print(key)

print("\n")

fields = [
    "admUnitCount",
    "boundaryID",
    "boundaryName",
    "boundaryISO",
    "boundaryType",
    "boundaryYearRepresented",
    "boundarySource",
    "boundarySourceURL",
    "boundaryLicense",
    "gjDownloadURL",
    "simplifiedGeometryGeoJSON",
]

for field in fields:
    print(field, ":", metadata.get(field))

print("\n")

boundary_url = "https://github.com/wmgeolab/geoBoundaries/raw/9469f09/releaseData/gbOpen/MOZ/ADM2/geoBoundaries-MOZ-ADM2.geojson"
print("Boundary download URL:", boundary_url)
```
