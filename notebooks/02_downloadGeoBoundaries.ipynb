```python
from pathlib import Path
import requests

boundary_url = "https://github.com/wmgeolab/geoBoundaries/raw/9469f09/releaseData/gbOpen/MOZ/ADM2/geoBoundaries-MOZ-ADM2.geojson"

out_dir = Path("../data/raw/boundaries")
out_dir.mkdir(parents=True, exist_ok=True)

out_path = out_dir / "geoBoundaries-MOZ-ADM2.geojson"

response = requests.get(boundary_url)
response.raise_for_status()

out_path.write_bytes(response.content)

print("Saved to:", out_path)
print("Size MB:", out_path.stat().st_size / 1_000_000)
```
