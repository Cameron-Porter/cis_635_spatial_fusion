# 2. Download Mozambique ADM2 boundaries

This notebook downloads the full geoBoundaries Mozambique ADM2 GeoJSON into `data/raw/boundaries/`. Raw downloaded data should be treated as source data and not manually edited.


## Load project paths and download URL


```python
from pathlib import Path
import sys

NOTEBOOK_DIR = Path.cwd()
PROJECT_ROOT = NOTEBOOK_DIR.parent if NOTEBOOK_DIR.name == "notebooks" else NOTEBOOK_DIR
SRC_DIR = PROJECT_ROOT / "src"
if str(SRC_DIR) not in sys.path:
    sys.path.append(str(SRC_DIR))

import requests

from childreach.paths import (
    ADM2_BOUNDARIES_GEOJSON,
    BOUNDARIES_RAW_DIR,
    GEBOUNDARIES_ADM2_GEOJSON_URL,
)

BOUNDARIES_RAW_DIR.mkdir(parents=True, exist_ok=True)

response = requests.get(GEBOUNDARIES_ADM2_GEOJSON_URL, timeout=60)
response.raise_for_status()

ADM2_BOUNDARIES_GEOJSON.write_bytes(response.content)

print("Saved to:", ADM2_BOUNDARIES_GEOJSON)
print("Size MB:", ADM2_BOUNDARIES_GEOJSON.stat().st_size / 1_000_000)
```

## Reproducibility note

The download URL is intentionally explicit. If geoBoundaries releases a newer file later, this notebook still records the exact source used for Milestone 1.
