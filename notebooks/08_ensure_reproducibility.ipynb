# 8. Reload processed Milestone 1 outputs

This notebook verifies that the saved GeoPackage and CSV can be loaded independently of the notebook that created them.


## Load processed outputs from disk


```python
from pathlib import Path
import sys

NOTEBOOK_DIR = Path.cwd()
PROJECT_ROOT = NOTEBOOK_DIR.parent if NOTEBOOK_DIR.name == "notebooks" else NOTEBOOK_DIR
SRC_DIR = PROJECT_ROOT / "src"
if str(SRC_DIR) not in sys.path:
    sys.path.append(str(SRC_DIR))

import geopandas as gpd
import pandas as pd

from childreach.paths import (
    MOZ_ADM2_UNDER18_2019_CSV,
    MOZ_ADM2_UNDER18_2019_GPKG,
    MOZ_ADM2_UNDER18_2019_LAYER,
)

gpkg_path = MOZ_ADM2_UNDER18_2019_GPKG
csv_path = MOZ_ADM2_UNDER18_2019_CSV

processed_gdf = gpd.read_file(gpkg_path, layer=MOZ_ADM2_UNDER18_2019_LAYER)
processed_csv = pd.read_csv(csv_path)

print(processed_gdf.shape)
print(processed_csv.shape)
print(processed_gdf.crs)
print(processed_gdf[["shapeName", "under18_sum", "valid_cell_count", "worldpop_valid_cells"]].head())
print(processed_gdf["worldpop_valid_cells"].value_counts(dropna=False))
```

## Check expected Milestone 1 conditions

These assertions make the reproducibility check explicit instead of relying only on visual inspection.


```python
assert len(processed_gdf) == 159
assert len(processed_csv) == 159
assert processed_gdf.crs.to_string() == "EPSG:4326"
assert processed_gdf["shapeID"].is_unique
assert processed_gdf["worldpop_valid_cells"].sum() == 157
assert (~processed_gdf["worldpop_valid_cells"]).sum() == 2

print("Milestone 1 processed outputs reload successfully.")
```

## Interpretation

The processed files are reusable by later notebooks, reports, and a possible app/demo. Future milestones should load these outputs from `data/processed/` rather than depending on live variables from previous notebooks.
