# 9. Create first ChildReach choropleth map

This notebook maps the processed ADM2 under-18 population estimates from Milestone 1.


## Load processed spatial output


```python
from pathlib import Path
import sys

NOTEBOOK_DIR = Path.cwd()
PROJECT_ROOT = NOTEBOOK_DIR.parent if NOTEBOOK_DIR.name == "notebooks" else NOTEBOOK_DIR
SRC_DIR = PROJECT_ROOT / "src"
if str(SRC_DIR) not in sys.path:
    sys.path.append(str(SRC_DIR))

import geopandas as gpd
import matplotlib.pyplot as plt

from childreach.paths import (
    FIGURES_DIR,
    MOZ_ADM2_UNDER18_2019_CHOROPLETH,
    MOZ_ADM2_UNDER18_2019_GPKG,
    MOZ_ADM2_UNDER18_2019_LAYER,
)

gpkg_path = MOZ_ADM2_UNDER18_2019_GPKG
processed_gdf = gpd.read_file(gpkg_path, layer=MOZ_ADM2_UNDER18_2019_LAYER)

print(processed_gdf.shape)
print(processed_gdf["worldpop_valid_cells"].value_counts(dropna=False))
```

## Render and save choropleth

Districts with no valid WorldPop raster support are intentionally shown with a separate missing-data style instead of being converted to zero.


```python
FIGURES_DIR.mkdir(parents=True, exist_ok=True)
figure_path = MOZ_ADM2_UNDER18_2019_CHOROPLETH

fig, ax = plt.subplots(figsize=(10, 12))

processed_gdf.plot(
    column="under18_sum",
    ax=ax,
    legend=True,
    cmap="OrRd",
    edgecolor="white",
    linewidth=0.2,
    missing_kwds={
        "color": "lightgray",
        "edgecolor": "red",
        "hatch": "///",
        "label": "No valid WorldPop cells",
    },
)

ax.set_title(
    "Estimated Population Under Age 18 by ADM2 District, Mozambique, 2019",
    fontsize=14,
)
ax.axis("off")

plt.tight_layout()
plt.savefig(figure_path, dpi=300, bbox_inches="tight")
plt.show()

print(figure_path)
print(figure_path.exists(), figure_path.stat().st_size)
```

## Interpretation

This map is a visual summary of Milestone 1, not a decision-making tool. It shows where modeled under-18 population is concentrated by ADM2 district and preserves the two no-raster-support islands as missing data.
