```python
from pathlib import Path
import geopandas as gpd
import matplotlib.pyplot as plt

gpkg_path = Path("../data/processed/moz_adm2_under18_2019.gpkg")
processed_gdf = gpd.read_file(gpkg_path, layer="moz_adm2_under18_2019")

figures_dir = Path("../figures")
figures_dir.mkdir(parents=True, exist_ok=True)

figure_path = figures_dir / "moz_adm2_under18_2019_choropleth.png"

fig, ax = plt.subplots(figsize=(10, 12))

processed_gdf.plot(
    column="under18_sum",
    ax=ax,
    legend=True,
    cmap="OrRd",
    edgecolor="white",
    linewidth=0.2,
    missing_kwds={
        "color": "lightgray",
        "edgecolor": "red",
        "hatch": "///",
        "label": "No valid WorldPop cells",
    },
)

ax.set_title(
    "Estimated Population Under Age 18 by ADM2 District, Mozambique, 2019",
    fontsize=14,
)

ax.axis("off")

plt.tight_layout()
plt.savefig(figure_path, dpi=300, bbox_inches="tight")
plt.show()

print(figure_path)
print(figure_path.exists(), figure_path.stat().st_size)
```
