```python
import geopandas as gpd

boundaries = gpd.read_file("../data/raw/boundaries/geoBoundaries-MOZ-ADM2.geojson")

print("Rows:", len(boundaries))
print("CRS:", boundaries.crs)

print("Missing geometries:", boundaries.geometry.isna().sum())
print("Invalid geometries:", (~boundaries.geometry.is_valid).sum())

print("Unique shapeIDs:", boundaries["shapeID"].nunique())
print("Duplicate shapeIDs:", boundaries["shapeID"].duplicated().sum())

print("Missing shapeNames:", boundaries["shapeName"].isna().sum())
print("Unique shapeNames:", boundaries["shapeName"].nunique())

print("shapeType values:")
print(boundaries["shapeType"].value_counts())

print("shapeGroup values:")
print(boundaries["shapeGroup"].value_counts())
```
