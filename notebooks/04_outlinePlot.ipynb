# 4. Preview ADM2 boundary outlines

This notebook creates a simple outline map. It is not the final analysis map; it is a visual sanity check that the boundary file loaded correctly and resembles Mozambique.


## Load boundaries and create preview figure


```python
from pathlib import Path
import sys

NOTEBOOK_DIR = Path.cwd()
PROJECT_ROOT = NOTEBOOK_DIR.parent if NOTEBOOK_DIR.name == "notebooks" else NOTEBOOK_DIR
SRC_DIR = PROJECT_ROOT / "src"
if str(SRC_DIR) not in sys.path:
    sys.path.append(str(SRC_DIR))

import geopandas as gpd
import matplotlib.pyplot as plt

from childreach.paths import ADM2_BOUNDARIES_GEOJSON, FIGURES_DIR, MOZ_ADM2_BOUNDARIES_PREVIEW

boundaries = gpd.read_file(ADM2_BOUNDARIES_GEOJSON)
FIGURES_DIR.mkdir(parents=True, exist_ok=True)

fig, ax = plt.subplots(figsize=(8, 10))
boundaries.plot(ax=ax, facecolor="none", edgecolor="black", linewidth=0.4)

ax.set_title("Mozambique ADM2 Boundaries - geoBoundaries")
ax.set_axis_off()

fig.savefig(MOZ_ADM2_BOUNDARIES_PREVIEW, dpi=200, bbox_inches="tight")
plt.show()

print(MOZ_ADM2_BOUNDARIES_PREVIEW)
print(MOZ_ADM2_BOUNDARIES_PREVIEW.exists(), MOZ_ADM2_BOUNDARIES_PREVIEW.stat().st_size)
```

## Interpretation

This preview confirms the vector layer is spatially recognizable before we use it as the zonal-statistics boundary layer.
